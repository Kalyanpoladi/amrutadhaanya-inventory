```python
def explore_column(df, keyword):
    """Finds and explores any column containing `keyword` (case-insensitive)."""
    matches = [c for c in df.columns if keyword.lower() in c.lower()]
    if not matches:
        print(f"No column containing '{keyword}'. Columns are:")
        print(list(df.columns))
        return
    for col in matches:
        print(f"\n📌 Column: '{col}' | dtype: {df[col].dtype}")
        print("Unique values:", df[col].unique()[:15])
        print(df[col].value_counts().head(10))

explore_column(df, 'status')
```

```python
dash_df = basket_df[[
    'Item_Name', 'Category', 'Unit',
    'Customer  Selling Price',      # note the double space!
    'Total_Qty', 'Remaining_Qty', 'Stock_Status'
]].copy()

# Rename for easy use (fixes the double-space problem forever)
dash_df = dash_df.rename(columns={'Customer  Selling Price': 'Selling_Price'})

# Drop duplicates on item name (keep first)
dash_df = dash_df.drop_duplicates(subset='Item_Name', keep='first')

# Sanity check: does price/qty make sense?
print("Negative prices:", (dash_df['Selling_Price'] < 0).sum())
print("Zero prices:", (dash_df['Selling_Price'] == 0).sum())
print("Negative qty:", (dash_df['Remaining_Qty'] < 0).sum())
print("Rows after dedup:", len(dash_df))

dash_df.head()
```

```python
print("="*55)
print("1. SHAPE")
print(f"Rows: {len(basket_df)} | Columns: {len(basket_df.columns)}")

print("="*55)
print("2. MISSING VALUES (dashboard columns only)")
cols = ['Item_Name','Category','Unit','Customer  Selling Price',
        'Total_Qty','Individual_Reserved_Qty','Basket_Reserved_Qty',
        'Remaining_Qty','Stock_Status','Added_Date']
print(basket_df[cols].isnull().sum())

print("="*55)
print("3. DUPLICATES")
print("Fully duplicate rows:", basket_df.duplicated().sum())
print("Duplicate Item_Names:", basket_df['Item_Name'].duplicated().sum())
dups = basket_df[basket_df['Item_Name'].duplicated(keep=False)]
if len(dups): display(dups[['Item_Name','Total_Qty','Customer  Selling Price','Added_Date']])

print("="*55)
print("4. PRICE SANITY")
p = basket_df['Customer  Selling Price']
print("Min:", p.min(), "| Max:", p.max(), "| Mean:", round(p.mean(),1))
print("Zero prices:", (p==0).sum(), "| Negative:", (p<0).sum())
# outlier suspects
display(basket_df.nlargest(5, 'Customer  Selling Price')[['Item_Name','Unit','Customer  Selling Price']])

print("="*55)
print("5. QTY LOGIC: Total == Individual + Basket + Remaining?")
bad = basket_df[basket_df['Total_Qty'] !=
      (basket_df['Individual_Reserved_Qty'] + basket_df['Basket_Reserved_Qty'] + basket_df['Remaining_Qty'])]
print(f"Rows breaking the rule: {len(bad)}")
if len(bad): display(bad[['Item_Name','Total_Qty','Individual_Reserved_Qty','Basket_Reserved_Qty','Remaining_Qty','Stock_Status']])

print("="*55)
print("6. UNIT CONSISTENCY (key for your dashboard!)")
print(basket_df['Unit'].value_counts())
# same item with different units?
mixed = basket_df.groupby('Item_Name')['Unit'].nunique()
print("Items sold in multiple units:", (mixed>1).sum())

print("="*55)
print("7. STOCK_STATUS vs REMAINING_QTY mismatch")
mismatch = basket_df[((basket_df['Stock_Status']=='Available') & (basket_df['Remaining_Qty']==0)) |
                     ((basket_df['Stock_Status']=='Out Of Stock') & (basket_df['Remaining_Qty']>0))]
print(f"Mismatched rows: {len(mismatch)}")
if len(mismatch): display(mismatch[['Item_Name','Remaining_Qty','Stock_Status']])

print("="*55)
print("8. CATEGORY TYPOS")
print(basket_df['Category'].value_counts())

print("="*55)
print("9. DATE RANGE")
print(basket_df['Added_Date'].min(), "→", basket_df['Added_Date'].max())
```

```python
clean = basket_df.copy()

# FIX 1: rename the double-space column
clean = clean.rename(columns={'Customer  Selling Price': 'Selling_Price'})

# FIX 2: impossible qty — Remaining can never exceed Total
mask = clean['Remaining_Qty'] > clean['Total_Qty']
print("Fixing impossible rows:", clean.loc[mask, 'Item_Name'].tolist())
clean.loc[mask, 'Remaining_Qty'] = clean.loc[mask, 'Total_Qty']   # cap at Total

# FIX 3: flag stale stock rows (Out of Stock with qty > 0)
clean['Stock_Flag'] = 'OK'
clean.loc[clean['Stock_Status'] == 'Out Of Stock', 'Stock_Flag'] = 'Stale - verify with team'

# FIX 4: effective qty available for dashboard
# Out-of-stock items show as 0 available, regardless of stale numbers
clean['Available_Qty'] = clean['Remaining_Qty']
clean.loc[clean['Stock_Status'] == 'Out Of Stock', 'Available_Qty'] = 0

# FIX 5: keep only dashboard columns, drop the empty ones
clean = clean[['Inventory_ID','Item_Name','Category','Unit','Selling_Price',
               'Total_Qty','Available_Qty','Stock_Status','Stock_Flag','Added_Date']]

# Final verification
print("\nFinal shape:", clean.shape)
print(clean['Stock_Flag'].value_counts())
display(clean.sample(5))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# ---- Chart 1: Potential Revenue by Category ----
revenue = pd.read_sql("""
    SELECT Category, SUM(Selling_Price * Available_Qty) AS Revenue
    FROM basket_inventory GROUP BY Category ORDER BY Revenue DESC
""", engine)

plt.figure(figsize=(10, 5))
sns.barplot(data=revenue, x='Revenue', y='Category', palette='viridis')
plt.title('Potential Revenue by Category (₹)')
plt.tight_layout()
plt.savefig('revenue_by_category.png', dpi=150)
plt.show()

# ---- Chart 2: Restock Urgency ----
urgent = pd.read_sql("""
    SELECT Item_Name, Selling_Price, Available_Qty
    FROM basket_inventory
    WHERE Available_Qty > 0 AND Available_Qty <= 10 AND Selling_Price >= 100
    ORDER BY Selling_Price / Available_Qty DESC
""", engine)

plt.figure(figsize=(10, 5))
sns.barplot(data=urgent, x='Available_Qty', y='Item_Name', palette='Reds_r')
plt.title('Restock Urgency: Low Stock, High Value Items')
plt.xlabel('Units Remaining')
plt.tight_layout()
plt.savefig('restock_urgency.png', dpi=150)
plt.show()

# ---- Chart 3: Price vs Quantity (Premium vs Bulk products) ----
all_items = pd.read_sql("SELECT Selling_Price, Available_Qty, Category FROM basket_inventory", engine)

plt.figure(figsize=(9, 6))
sns.scatterplot(data=all_items, x='Available_Qty', y='Selling_Price', hue='Category', s=80)
plt.title('Product Positioning: Price vs Stock')
plt.tight_layout()
plt.savefig('price_vs_qty.png', dpi=150)
plt.show()
```

```python
with pd.ExcelWriter("veggies_fruits_millets.xlsx", engine="xlsxwriter") as writer:
    filtered.to_excel(writer, index=False, sheet_name="Price List")
    workbook  = writer.book
    worksheet = writer.sheets["Price List"]

    # Format
    money = workbook.add_format({'num_format': '₹#,##0.00'})
    worksheet.set_column('A:A', 40)   # Item_Name wide
    worksheet.set_column('B:B', 15, money)  # Price as ₹
    worksheet.set_column('C:E', 15)

    # Total row at bottom
    total_row = len(filtered) + 1
    worksheet.write(total_row, 0, "TOTAL VALUE")
    worksheet.write(total_row, 1, filtered['Selling_Price'].mul(filtered['Total_Qty']).sum(), money)

print("Formatted Excel saved ✅")
```
